```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from src.llm.prompts.writing_task_2.lexical_resources.vocab_range import vocab_range_prompt , parser

load_dotenv()
```

```python
essay = """
Topic : Governments in many countries spend large amounts of money on space exploration. Some people believe this money should instead be spent solving problems on Earth. To what extent do you agree or disagree?

Sample: it is argued that in the large number of countries, governments allocated large amounts of financial resources to space exploration. Some individuals suggest that it is essential to spend budget to address issues on earth. From my perspective, I partly agree with this viewpoint due to several reasons that will be explained in the following essay.

On the one hand, the major reason that explain why many countries spend a great deal of money and effort on space exploration is innovation and development. When financial investment in space exploration is more and more, scientists have sufficient resources to generate science breakthroughs such as renewable materials and new technology that can change and improve the living standards. As a result, this investment can serve as the primary motivation, which often promotes the development, potentially helping solve problems in the future.

However, there are several compelling reasons why some people advocate for focusing on tackling problems on earth is poverty and pollution. Nowadays, many people are suffering from the lack of access to essentials such as food and clear water, which is limited by industrial manufacturing excessively, and industrial expansion, especially in many Africa countries.
As a result, providing financial assistance to them can leads to poverty alleviation and improve living standards better instead of investing in space exploration, which might take a large mount of time to achieve.

In conclusion, while exploring space can bring benefits in the long run in terms of innovation, I truly believe that financially focusing more on addressing on earth is more beneficial.
"""
```

```python
prompt_value = vocab_range_prompt.invoke({
    "essay": essay,
    "format_instructions": parser.get_format_instructions()
})

# Để xem nội dung text từ PromptValue:
print(prompt_value.to_string())

# Hoặc xem dưới dạng danh sách tin nhắn (nếu là Chat Model):
print(prompt_value.to_messages())
```

```python
llm = ChatOpenAI(name='gpt-5-mini', temperature=0.0)

format_instructions = parser.get_format_instructions()

chain = vocab_range_prompt | llm | parser

res = chain.invoke({
    'format_instructions' : format_instructions,
    'essay' : essay
})
```

```python
print(res.vocab_table_analysis)
print("---")
print(res.summary)
```
